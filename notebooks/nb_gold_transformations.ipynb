```python
from pyspark.sql.functions import *

orders = spark.table(
    "silver_orders"
)

gold_customer_lifetime_value = (
    orders
    .groupBy("customer_id")
    .agg(
        sum("total_amount").alias(
            "lifetime_value"
        ),
        count("order_id").alias(
            "total_orders"
        )
    )
)

gold_customer_lifetime_value.write \
    .mode("overwrite") \
    .format("delta") \
    .saveAsTable(
        "gold_customer_lifetime_value"
    )
```

```python
# validation
display(
    gold_customer_lifetime_value
)
```

```python
# sales metrics
gold_sales_metrics = (
    orders
    .groupBy("sales_channel")
    .agg(
        sum("total_amount")
            .alias("revenue"),

        count("order_id")
            .alias("orders")
    )
)

gold_sales_metrics.write \
    .mode("overwrite") \
    .format("delta") \
    .saveAsTable(
        "gold_sales_metrics"
    )
```

```python
# validation
display(gold_sales_metrics)
```

```python
# product performance
order_items = spark.table(
    "silver_order_items"
)

products = spark.table(
    "silver_products"
)

gold_product_performance = (
    order_items
    .join(
        products,
        "product_id"
    )
    .groupBy(
        "product_id",
        "product_name",
        "category"
    )
    .agg(
        sum("quantity")
        .alias("units_sold")
    )
)
```

```python
# save
gold_product_performance.write \
    .mode("overwrite") \
    .format("delta") \
    .saveAsTable(
        "gold_product_performance"
    )
```

```python
# return analysis
returns = spark.table(
    "silver_returns"
)

gold_returns_analysis = (
    returns
    .groupBy("return_reason")
    .agg(
        count("*")
            .alias("total_returns"),

        sum("refund_amount")
            .alias("refund_value")
    )
)
```

```python
# save
gold_returns_analysis.write \
    .mode("overwrite") \
    .format("delta") \
    .saveAsTable(
        "gold_returns_analysis"
    )
```
