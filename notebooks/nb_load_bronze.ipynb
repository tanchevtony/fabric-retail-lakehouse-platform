```python
# Read customer CSV

df = spark.read.csv(
    "Files/bronze/customers.csv",
    header=True,
    inferSchema=True
)

df.write.mode("overwrite") \
    .format("delta") \
    .saveAsTable("bronze_customers")
```

```python
# validate bronze customers
spark.sql("""
SELECT COUNT(*)
FROM bronze_customers
""").show()
```

```python
display(
    spark.table("bronze_customers")
)
```

```python
# Read products,stores, orders, order_items, returns CSV
tables = [
    "products",
    "stores",
    "orders",
    "order_items",
    "returns"
]

for table in tables:

    df = spark.read.csv(
        f"Files/bronze/{table}.csv",
        header=True,
        inferSchema=True
    )

    df.write.mode("overwrite") \
      .format("delta") \
      .saveAsTable(f"bronze_{table}")

    print(f"Loaded bronze_{table}")
```

```python
# validate bronze products
spark.sql("""
SELECT COUNT(*)
FROM bronze_products
""").show()
```

```python
# validate bronze stores
spark.sql("""
SELECT COUNT(*)
FROM bronze_stores
""").show()
```

```python
spark.sql("""
SELECT COUNT(*)
FROM bronze_orders
""").show()
```

```python
# validate bronze order-items
spark.sql("""
SELECT COUNT(*)
FROM bronze_order_items
""").show()
```

```python
spark.sql("""
SELECT COUNT(*)
FROM bronze_returns
""").show()
```
